```python
import cv2  # Qué: importa OpenCV, que incluye los algoritmos de detección de características (ORB, SIFT) usados en este notebook.
```

```python
image = cv2.imread("../imgs/img2.jpg")  # Qué: carga la imagen a color sobre la que se van a detectar y dibujar los puntos clave.
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)  # Qué: convierte la imagen a escala de grises. Por qué: los detectores de características como ORB y SIFT analizan gradientes/intensidad, no información de color, así que trabajan sobre una sola banda.
```

```python
orb = cv2.ORB_create()  # Qué: instancia el detector/descriptor ORB (Oriented FAST and Rotated BRIEF) con sus parámetros por defecto. Por qué: ORB es rápido, libre de patentes y funciona bien en tiempo real, aunque es menos preciso que SIFT ante cambios de escala grandes.

sift = cv2.SIFT_create()  # Qué: instancia el detector/descriptor SIFT (Scale-Invariant Feature Transform) con parámetros por defecto. Por qué: SIFT es más robusto a cambios de escala, rotación e iluminación que ORB, a costa de ser más lento; se compara ambos para ver el trade-off precisión/velocidad.
```

# Detectar los puntos clave y calcular los descriptores

```python
keypointsORB, descriptorsORB =   orb.detectAndCompute(gray_image, None)  # Qué: detecta puntos clave (esquinas/regiones distintivas) y calcula su descriptor binario con ORB en un solo paso. Cómo: detectAndCompute recibe la imagen en gris y una máscara opcional (None = analizar toda la imagen); devuelve la lista de keypoints y la matriz de descriptores asociada.

keypointsSIFT, descriptorsSIFT = sift.detectAndCompute(gray_image, None)  # Qué: igual que arriba pero con SIFT, que genera descriptores de punto flotante de 128 dimensiones (más ricos que los binarios de ORB, pero más pesados de comparar).
```

## Detectar los puntos clave y calcular los descriptores

```python
image_with_keypointsORB = cv2.drawKeypoints(image, keypointsORB, None , color=(0,255,0))  # Qué: dibuja los puntos clave de ORB sobre una copia de la imagen original a color. Cómo: el tercer argumento None indica que no se reutiliza un array de salida existente (se crea uno nuevo); color=(0,255,0) pinta los marcadores en verde (BGR).

image_with_keypointsSIFT = cv2.drawKeypoints(image, keypointsSIFT, None, color=(0,255,0))  # Qué: mismo dibujo pero con los keypoints detectados por SIFT, para poder comparar visualmente la cantidad y ubicación de puntos frente a ORB.
```

## Dibujar los puntos clave en la imagen

```python
cv2.imshow("Puntos Clave SIFT", image_with_keypointsSIFT)  # Qué: abre una ventana con los puntos clave de SIFT dibujados.

cv2.imshow("Puntos Clave ORB", image_with_keypointsORB)  # Qué: abre otra ventana con los puntos clave de ORB, para comparar lado a lado con SIFT.

cv2.waitKey(0)  # Qué: espera indefinidamente a que se presione una tecla antes de cerrar las ventanas.
cv2.destroyAllWindows()  # Qué: cierra ambas ventanas para liberar los recursos de GUI.
```
