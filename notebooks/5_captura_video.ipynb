```python
import cv2  # Qué: importa el módulo principal de OpenCV. Cómo: expone las funciones de captura, transformación de color y GUI (imshow/waitKey) usadas en este notebook. Por qué: es la única dependencia externa necesaria para capturar y mostrar video.
```

```python
video = cv2.VideoCapture(0)  # Qué: abre un flujo de captura desde la cámara con índice 0 (la webcam por defecto). Cómo: usa el backend de captura del SO (DirectShow/MSMF en Windows). Por qué: 0 es la convención de OpenCV para "el primer dispositivo de video detectado".

while True:  # Qué: bucle infinito de lectura de video. Por qué: una cámara no tiene "fin" natural, así que se procesa frame a frame hasta que el usuario decida salir (con la tecla 'q' más abajo).

    ret, frame = video.read()  # Qué: lee el siguiente cuadro del flujo. Cómo: devuelve una tupla (ret, frame): `ret` es un booleano que indica si la lectura fue exitosa y `frame` es la imagen como array de NumPy en formato BGR (el orden de canales nativo de OpenCV).

    hsv_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)  # Qué: convierte el cuadro de BGR a HSV (Matiz, Saturación, Valor). Cómo: aplica la fórmula de conversión de color de OpenCV con el flag COLOR_BGR2HSV. Por qué: HSV separa el color (matiz) de la iluminación (valor), lo que facilita tareas como segmentación por color en módulos posteriores.


    cv2.imshow("Cuadro a Color", frame)  # Qué: abre/actualiza una ventana con el cuadro original en BGR. Cómo: crea una ventana con el título dado si no existe y refresca su contenido. Por qué: permite comparar visualmente el video original contra el filtrado.

    cv2.imshow("Cuadro con filtro HSV", hsv_frame)  # Qué: muestra en otra ventana el cuadro convertido a HSV. Por qué: evidencia visualmente el efecto de la conversión de espacio de color.

    if cv2.waitKey(1) == ord('q'):  # Qué: espera 1 ms por una tecla y compara su código con el de 'q'. Cómo: waitKey también procesa los eventos de las ventanas de GUI (sin esto, imshow no refresca). Por qué: waitKey(1) mantiene el video fluido (bajo delay) mientras permite salir del loop presionando 'q'.
        break  # Qué: corta el bucle infinito. Por qué: es la única forma de terminar la captura de forma controlada.
video.release()  # Qué: libera el dispositivo de captura. Cómo: cierra el handle del SO hacia la cámara. Por qué: si no se libera, otros procesos (u otra celda) no podrán acceder a la webcam.
cv2.destroyAllWindows()  # Qué: cierra todas las ventanas de OpenCV abiertas. Por qué: evita ventanas "colgadas" que no responden tras terminar el script.
```
