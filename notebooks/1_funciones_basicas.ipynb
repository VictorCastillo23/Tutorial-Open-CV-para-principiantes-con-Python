```python
import cv2  # Qué: importa el módulo de OpenCV para Python. Cómo: el paquete se instala como "opencv-python" pero se importa con el alias "cv2" por compatibilidad histórica con la API de C++. Por qué: todas las funciones de visión por computadora del tutorial (lectura, transformación, dibujo) viven en este módulo.
```

# Lectura de imagen a color
Carga una imagen desde disco conservando sus canales de color originales (BGR)

```python
image1 = cv2.imread("../imgs/img1.jpg")  # Qué: lee el archivo JPG del disco y lo decodifica a una matriz NumPy. Cómo: cv2.imread devuelve un array de shape (alto, ancho, 3) sin el flag de modo (por defecto lee a color). Por qué: OpenCV usa el orden de canales BGR (no RGB) por una decisión histórica de la librería original en C++.
```

```python
cv2.imshow("Imagen en Color", image1)  # Qué: abre una ventana nativa con el título indicado y dibuja la matriz image1. Cómo: cv2.imshow usa el backend de GUI de OpenCV (HighGUI); no funciona directo en notebooks sin backend gráfico, por eso se ejecuta como script/ventana aparte.

cv2.waitKey(0)  # Qué: pausa la ejecución esperando una tecla. Cómo: el argumento 0 significa "esperar indefinidamente" (si fuera >0 sería milisegundos). Por qué: sin este llamado la ventana se abriría y cerraría instantáneamente al terminar el script.
cv2.destroyAllWindows()  # Qué: cierra todas las ventanas de OpenCV abiertas. Por qué: libera los recursos de GUI una vez que el usuario ya vio la imagen.
```

# Lectura de imagen en escala de grises
Carga la imagen en escala de grises pasando el flag `0` como segundo argumento de `cv2.imread`

```python
image2 = cv2.imread("../imgs/img2.jpg",0)  # Qué: lee la imagen y la decodifica directo en escala de grises. Cómo: el segundo argumento 0 equivale al flag cv2.IMREAD_GRAYSCALE, así el array resultante tiene shape (alto, ancho) con un solo canal. Por qué: decodificar en gris de una vez es más eficiente que leer a color y convertir después con cv2.cvtColor.

cv2.imshow("Imagen con filtro", image2)  # Qué: muestra la imagen en gris en una ventana nueva.
cv2.waitKey(0)  # Qué: espera indefinidamente a que se presione una tecla antes de continuar.
cv2.destroyAllWindows()  # Qué: cierra todas las ventanas abiertas para liberar los recursos de GUI.
```

# Redimensionado con factor de escala
Cambia el tamaño de la imagen aplicando un factor de escala sobre cada eje con los parámetros `fx` y `fy`

```python
scaled_image = cv2.resize(image1, (0, 0), fx=0.5, fy=0.5)  # Qué: reduce la imagen a la mitad de su tamaño original. Cómo: al pasar (0, 0) como tamaño destino, cv2.resize ignora ese argumento y calcula las nuevas dimensiones multiplicando alto y ancho por fx/fy. Por qué: usar un factor de escala relativo evita tener que calcular a mano las dimensiones exactas y preserva la relación de aspecto.
```

# Redimensionado a tamaño fijo
Cambia el tamaño de la imagen a unas dimensiones exactas en píxeles, sin conservar la relación de aspecto original

```python
scaled_image = cv2.resize(image1, (500,500))  # Qué: fuerza la imagen a 500x500 píxeles exactos. Cómo: al pasar una tupla (ancho, alto) explícita, cv2.resize interpola los píxeles al nuevo tamaño sin usar fx/fy. Por qué: útil cuando se necesita un tamaño fijo (por ejemplo, para un modelo de ML), aunque distorsiona la imagen si la relación de aspecto original no era cuadrada.
```

# Mostrar imagen escalada
Muestra en una ventana el resultado del redimensionado

```python
cv2.imshow("Imagen Escalada", scaled_image)  # Qué: muestra la imagen redimensionada en una ventana.
cv2.waitKey(0)  # Qué: bloquea la ejecución hasta que se presione una tecla.
cv2.destroyAllWindows()  # Qué: cierra la ventana para liberar los recursos de GUI.
```

# Rotación de imagen
Calcula la matriz de rotación alrededor del centro de la imagen y la aplica con `cv2.warpAffine`

```python
(height, width) = image1.shape[:2]  # Qué: extrae alto y ancho de la imagen. Cómo: image1.shape devuelve (alto, ancho, canales); [:2] descarta el número de canales, que no hace falta para calcular el centro de rotación.
center = (width // 2, height // 2)  # Qué: calcula el punto central de la imagen en coordenadas (x, y). Cómo: división entera (//) para obtener un índice de píxel válido. Por qué: rotar alrededor del centro (y no de la esquina) evita que la imagen "salte" fuera de cuadro de forma asimétrica.
	
rotation_matrix = cv2.getRotationMatrix2D(center, 70, 1.0)  # Qué: construye la matriz de transformación afín 2x3 para rotar. Cómo: recibe el punto pivote, el ángulo en grados (70°, sentido antihorario) y un factor de escala (1.0 = sin reescalar). Por qué: cv2.warpAffine necesita esta matriz explícita en vez de un ángulo, ya que trabaja de forma genérica con cualquier transformación afín.

rotated_image = cv2.warpAffine(image1, rotation_matrix, (width, height))  # Qué: aplica la matriz de rotación a la imagen. Cómo: warpAffine remapea cada píxel según la matriz y recorta/rellena al tamaño de salida indicado (width, height), que aquí es el mismo tamaño original. Por qué: al mantener el tamaño original, las esquinas de la imagen rotada pueden quedar recortadas.
```

# Mostrar imagen rotada
Muestra en una ventana el resultado de la rotación

```python
cv2.imshow("Imagen en Rotada", rotated_image)  # Qué: muestra la imagen ya rotada en una ventana.
cv2.waitKey(0)  # Qué: bloquea la ejecución hasta que se presione una tecla.
cv2.destroyAllWindows()  # Qué: cierra la ventana para liberar los recursos de GUI.
```
