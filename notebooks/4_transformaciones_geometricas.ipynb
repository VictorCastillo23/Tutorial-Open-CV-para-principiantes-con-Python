```python
import cv2  # Qué: importa OpenCV para leer la imagen y aplicar la transformación de perspectiva.
import numpy as np  # Qué: importa NumPy para construir los arrays de puntos (float32) que exige cv2.getPerspectiveTransform. Por qué: OpenCV requiere que las coordenadas de los puntos sean float32, no listas de Python planas.
```

```python
image = cv2.imread("../imgs/img2.jpg")  # Qué: carga la imagen a color sobre la que se va a aplicar el cambio de perspectiva.
```

# Definir cuatro puntos de la imagen original que forman un cuadrilátero

```python
pts1 = np.float32([[50, 50], [200, 50], [50, 200], [200, 200]])  # Qué: define 4 puntos (x, y) en la imagen de origen que forman un cuadrilátero de referencia. Cómo: se guardan como np.float32 porque cv2.getPerspectiveTransform exige coordenadas en punto flotante. Por qué: una transformación de perspectiva necesita 4 puntos de correspondencia (no 3, como una afín) para resolver la matriz homográfica 3x3.
```

# Definir la nueva ubicación de los puntos en la imagen transformada

```python
pts2 = np.float32([[10, 100], [200, 50], [100, 250], [290, 220]])  # Qué: define los 4 puntos de destino, es decir, dónde debe terminar cada punto de pts1 tras la transformación. Cómo: el orden de los puntos debe corresponder uno a uno con pts1 (primer punto de pts1 va al primer punto de pts2, etc.). Por qué: al mover estos puntos a posiciones no colineales/no rectangulares se logra el efecto de "estirar"/inclinar la imagen como si se viera desde otro ángulo.
```

# Obtener la matriz de transformación de perspectiva

```python
perspective_matrix = cv2.getPerspectiveTransform(pts1, pts2)  # Qué: calcula la matriz de homografía 3x3 que mapea pts1 a pts2. Cómo: resuelve el sistema de ecuaciones lineales que satisface la correspondencia entre ambos conjuntos de puntos. Por qué: a diferencia de una transformación afín (warpAffine), la de perspectiva puede simular profundidad/inclinación porque no preserva el paralelismo de las líneas.
```

```python
(height, width) = image.shape[:2]  # Qué: extrae alto y ancho de la imagen original para definir el tamaño del lienzo de salida.

other_perspective = cv2.warpPerspective(image, perspective_matrix, (height, width))  # Qué: aplica la matriz de perspectiva a la imagen. Cómo: warpPerspective remapea cada píxel según la homografía y genera una imagen del tamaño indicado en el tercer argumento. Nota: aquí se pasa (height, width) en vez de (width, height) como espera la API (ancho, alto); si la imagen no es cuadrada el resultado queda con proporciones distintas a las originales.
```

```python
cv2.imshow("Imagen con otra Perspectiva", other_perspective)  # Qué: muestra el resultado de la transformación de perspectiva en una ventana.
cv2.waitKey(0)  # Qué: espera indefinidamente a que se presione una tecla.
cv2.destroyAllWindows()  # Qué: cierra la ventana para liberar los recursos de GUI.
```
